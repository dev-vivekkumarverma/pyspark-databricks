```
class testStreamingWordCount(StreamProcessingWordCount):
    def __init__(self,project_dir:str):
        self.project_dir=project_dir
        self.landing_zone="landing_zone/"
        self.dataset_dir="dataset/"
        self.checkpoint_dir="checkpoint/"
        self.load_table_name="proj2_word_count_table"
        self.result_df=None
        self.streamingQuery=None

    def assert_data(self,expected:int):
        actual = spark.sql(f"select sum(count) from {self.load_table_name} where substr(word, 1,1)=='s'").collect()[0][0]
        assert expected == actual, f"test failed ! actual value: {actual}  and expected: {expected}"

    def runtest(self, file_name:str):
        streamingQuery=self.process_data(file_name)
        self.streamingQuery=streamingQuery
        print("TEST RUN RES: successfull !")
        
    def getStreamingQuery(self):
        return self.streamingQuery
    

```

- run the test cases

```
import time
```

```
fileName_expected_value_list=[("text_data_1.txt",25),
                            ("text_data_2.txt",32),
                            ("text_data_3.txt",37)]
sleep_time=30 #in seconds

StrWrdCntTest=testStreamingWordCount(project_dir=project_dir)

for i in range(0,3):
    fileName, expected_val=fileName_expected_value_list[i]
    if i==0:
        StrWrdCntTest.runtest(file_name=fileName)
    else:
        StrWrdCntTest.ingest_data(fileName=fileName)
    print("*"*3 ,f'sleeping for {sleep_time} sec', "*"*3)
    time.sleep(sleep_time)   
    StrWrdCntTest.assert_data(expected=expected_val)

    print("--"*10, f"Test {i+1} with param: filename: {fileName} & expected_val:{expected_val},  competed successfully", "--"*10)
streamingQuery=StrWrdCntTest.getStreamingQuery()
streamingQuery.stop()
stremStopRes=streamingQuery.status["message"]

if stremStopRes =="Stopped":
    print("##"*10," STREAMING STOPED SUCCESSFULLY !","##"*10)

```
