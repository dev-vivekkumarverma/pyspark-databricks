```
import time
```

```
class testCases():
    def __init__(self,project_dir:str):
        self.project_dir=project_dir
        self.bronzeObj=BronzeLayer(project_dir=self.project_dir)
        self.silverObj=SilverLayer(project_dir=self.project_dir)


    def ingest_file(self,fileName:str):
        self.bronzeObj.ingest_data_to_landing_zone(file_name=fileName)
        
    def startBronzeLayer(self):
        self.bronzeObj.cleanup_n_setup()
        self.bronzeObj.extract_data()
        self.bronzeObj.load_data()

    def stopBronzeLayer(self):
        self.bronzeObj.stopStreamingQuery()
    
    def startSilverLayer(self):
        self.silverObj.cleanup_n_setup()
        self.silverObj.extract_data()
        self.silverObj.transform_data()
        self.silverObj.load_data()

    def stopSilverLayer(self):
        self.silverObj.stopStream()
    
    # def assert_data_bronze(self, expected_val:int):
    #     row_count_query_bronze=f"select count(*) from {self.bronzeObj.getTableName()}"
    #     actualVal=spark.sql(row_count_query_bronze).collect()[0][0]
    #     print("BRONZE: ACTUAL VAL: ",actualVal, "Expected Val: ",expected_val)
    #     # assert expected_val==actualVal, f"test failed (bronze) ! Actual val: {actualVal}, Expected val: {expected_val}"
    #     print("TEST RES: PASSED !")

    def assert_data_silver(self, expected_val:int):
        row_count_query_silver=f"select count(*) from {self.silverObj.getTableName()}"
        actualVal=spark.sql(row_count_query_silver).collect()[0][0]
        print("SILVER: ACTUAL VAL: ",actualVal, "Expected Val: ",expected_val)
        assert expected_val==actualVal, f"test failed (silver) ! Actual val: {actualVal}, Expected val: {expected_val}"
        print("TEST RES: PASSED !")

    def runTest(self):
        sleep_time=40
        test_param_list=[("invoices_1.json",1253),("invoices_2.json",2510),("invoices_3.json",3994),("invoices_4.json",4016)]
        self.startBronzeLayer()
        self.startSilverLayer()
        # print(f"sleeping for {sleep_time} seconds !")
        # time.sleep(sleep_time)
        i=1
        for fileName, expectedVal in test_param_list:
            print("--"*10, f" RUNNING TEST CASE: {i} ", "--"*10)
            self.ingest_file(fileName=fileName)
            # print(f"sleeping for {sleep_time} seconds !")
            # time.sleep(sleep_time+10)
            # self.assert_data_bronze(expectedVal)
            print(f"sleeping for {sleep_time} seconds !")
            time.sleep(sleep_time)
            self.assert_data_silver(expectedVal)
            i+=1
           

        print("===="*10)
        print("ALL TEST COMPLETED !")
        print("SAMPLE DATA- BRONZE LAYER:")
        display(self.bronzeObj.getHeadData(30))

        print("SAMPLE DATA- SILVER LAYER:")
        display(self.silverObj.getHeadData(30))


        print("STOPPING THE STREAMING QUERIES !")
        self.stopBronzeLayer()
        self.stopSilverLayer()

            
        print("#####"*20)

        
        
```
