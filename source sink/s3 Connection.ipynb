- s3 connector using normal way (`without mouting the bucket`)

````python
access_key=''
secret_key=''
bucket_name=''
aws_region=''
spark.conf.set("fs.s3a.access.key", access_key)
spark.conf.set("fs.s3a.secret.key", secret_key)
spark.conf.set("fs.s3a.endpoint", "s3.amazonaws.com")

````

````python
df= spark.read.csv('s3a://'+bucket_name+"/*", inferSchema=True, header=True)
````

````python
mount_res=dbutils.fs.mount(
  source='s3a://'+bucket_name,
  mount_point="/mnt/"+bucket_name,
  extra_configs={
    'fs.s3a.access.key':access_key,
    'fs.s3a.seceret.key':secret_key,
    "fs.s3a.endpoint": "s3.amazonaws.com"
  }
)
if mount_res:
  print(bucket_name+' mounted successfully at /mnt/'+bucket_name)

````
