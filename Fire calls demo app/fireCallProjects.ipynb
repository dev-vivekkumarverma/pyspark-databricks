- Changing the dataframe to a view

```
fire_df.createGlobalTempView("fire_call_table")
```

now a view is created with the name Of `fire_call_table` in global_temp database
we can run SQL Query on it

```
%sql
select * from global_temp.fire_call_table limit 10;
```
