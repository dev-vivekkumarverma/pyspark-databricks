Q1. How many distict tyoes of calls were made to the fire department ?

````
%sql
select count(distinct(calltype)) from my_database.fire_call_table where calltype is not null;
````

q2. What were distinct tyoes of calls Made to the fire department ?

````
%sql
select distinct(calltype) from my_database.fire_call_table where CallType is not null;
````

Q3. Find out all respose from delayed times greater than 5 mins?

````
%sql
select callNumber, delay from my_database.fire_call_table where Delay >5
````

Q4. What were the most common call types ?

````
%sql
with temp as (select callType, count(*) as freq from my_database.fire_call_table group by CallType having calltype is not null) 
select calltype, freq from temp where freq=(select max(freq) from temp)
````

Q5. what zip codes accounted for most common calls ?

````
%sql
select calltype, ZipcodeofIncident as zip, count(*) as total_call_incidents from my_database.fire_call_table where CallType is not null 
group by CallType, zip 
order by total_call_incidents desc 
-- limit 1
````
