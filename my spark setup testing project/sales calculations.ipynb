```
from pyspark.sql.functions import avg,sum
result=df2.groupBy(df2.payment_status).agg(sum(df2.product_price).alias("sales"))
```

```
display(result)
```

```
result1=result.orderBy([result.sales] ,ascending=[False])
```

```
display(result1)
```

```
joineddf= df1.join(other=df2, on="merchant_id")
```

```
display(joineddf)
```

```
product_sales_per_city=joineddf.groupBy([joineddf.city,joineddf.payment_status]).agg(sum(joineddf.product_price).alias("total_sales")).orderBy([joineddf.city,joineddf.payment_status],ascending=[True,True])
```
